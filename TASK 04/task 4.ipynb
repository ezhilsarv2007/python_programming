```python
import math

# Given values
fat = 4.2
snf = 3.1

# Constants
FAT_BASE = 3.5
SNF_BASE = 8.5
STEP = 0.1

BASE_RATE = 30
FAT_RATE = 2
SNF_RATE = 1.5
LITRES = 10

# 1. Calculate number of fat and SNF steps
fat_steps = round((fat - FAT_BASE) / STEP)
snf_steps = round((snf - SNF_BASE) / STEP)

print("Fat steps:", fat_steps)
print("SNF steps:", snf_steps)

# 2. Calculate rate per litre
rate_per_litre = BASE_RATE + (fat_steps * FAT_RATE) + (snf_steps * SNF_RATE)

print("Rate per litre:", rate_per_litre)

# 3. Calculate total amount
amount = LITRES * rate_per_litre

print("Total amount:", amount)

# 4. Comparison chaining
print("Valid fat:", 2.0 <= fat <= 7.0)

# 5. Boolean operators
print("Valid fat and SNF:", 2.0 <= fat <= 7.0 and snf >= 0)

# 6. Floating-point surprise
print("\nUnexpected Result 1:")
print("(4.2 - 3.5) / 0.1 =", (4.2 - 3.5) / 0.1)
print("Using round():", round((4.2 - 3.5) / 0.1))

# 7. Operator precedence surprise
print("\nUnexpected Result 2:")
print("-2 ** 2 =", -2 ** 2)
print("(-2) ** 2 =", (-2) ** 2)

# 8. Negative value difference
print("\nUnexpected Result 3:")
x = -2.7
print("round(-2.7):", round(x))
print("int(-2.7):", int(x))
print("math.floor(-2.7):", math.floor(x))

# 9. Alternative calculation for fat = 4.2
fat = 4.2

print("\nFor fat = 4.2:")
print("round():", round((fat - FAT_BASE) / STEP))
print("int():", int((fat - FAT_BASE) / STEP))
print("floor():", math.floor((fat - FAT_BASE) / STEP))

# 10. Alternative calculation for fat = 3.1
fat = 3.1

print("\nFor fat = 3.1:")
print("round():", round((fat - FAT_BASE) / STEP))
print("int():", int((fat - FAT_BASE) / STEP))
print("floor():", math.floor((fat - FAT_BASE) / STEP))

# 11. Exponentiation associativity
print("\nExponentiation:")
print("2 ** 3 ** 2 =", 2 ** 3 ** 2)
print("2 ** (3 ** 2) =", 2 ** (3 ** 2))
```
