```python
# TASK 6 - Organise the Logic into Functions

# Rate constants
BASE_RATE = 30
FAT_RATE = 2
SNF_RATE = 1.5
STEP = 0.1
FAT_BASE = 3.5
SNF_BASE = 8.5


def rate_per_litre(fat, snf):
    """Calculate and return the milk rate per litre."""
    fat_steps = round((fat - FAT_BASE) / STEP)
    snf_steps = round((snf - SNF_BASE) / STEP)

    rate = BASE_RATE + (fat_steps * FAT_RATE) + (snf_steps * SNF_RATE)

    return rate


def payout(litres, rate):
    """Calculate and return the payment for the given quantity of milk."""
    amount = litres * rate
    return amount


def ten_day_total(daily_amounts):
    """Calculate and return the total payout for ten days."""
    total = sum(daily_amounts)
    return total


# Read input
fat = float(input("Enter fat value: "))
snf = float(input("Enter SNF value: "))
litres = float(input("Enter litres: "))

# Function calls
rate = rate_per_litre(fat, snf)
amount = payout(litres, rate)

# Example ten-day amounts
daily_amounts = [amount] * 10
total = ten_day_total(daily_amounts)

# Display results
print("\nPayment Details")
print("-" * 35)
print(f"{'Rate per litre:':<20}{rate:8.2f}")
print(f"{'Daily payout:':<20}{amount:8.2f}")
print(f"{'Ten-day total:':<20}{total:8.2f}")
```
