```python
# TASK 5 - Read Input and Convert Types

print("Milk Payment Calculator")

# Read input
fat_text = input("Enter fat value: ")
snf_text = input("Enter SNF value: ")
litres_text = input("Enter litres: ")

# Remove extra spaces
fat_text = fat_text.strip()
snf_text = snf_text.strip()
litres_text = litres_text.strip()

# Check and convert input
try:
    fat = float(fat_text)
    snf = float(snf_text)
    litres = float(litres_text)

    print("\nInput accepted successfully.")

    # Example calculation
    fat_steps = round((fat - 3.5) / 0.1)
    snf_steps = round((snf - 8.5) / 0.1)

    base_rate = 30
    fat_rate = 2
    snf_rate = 1.5

    rate_per_litre = base_rate + (fat_steps * fat_rate) + (snf_steps * snf_rate)
    amount = litres * rate_per_litre

    # Aligned output using f-strings
    print("\nPayment Details")
    print("-" * 35)
    print(f"{'Fat:':<15}{fat:8.2f}")
    print(f"{'SNF:':<15}{snf:8.2f}")
    print(f"{'Litres:':<15}{litres:8.2f}")
    print(f"{'Rate/Litre:':<15}{rate_per_litre:8.2f}")
    print(f"{'Amount:':<15}{amount:8.2f}")

except ValueError:
    print("\nInvalid input!")
    print("Please enter numbers such as 8.5, not '8,5' or 'eight'.")
```
